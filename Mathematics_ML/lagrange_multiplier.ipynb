```python
import numpy as np

# First we define the functions,
def f (x, y) :
    return np.exp(-(2*x*x + y*y - x*y) / 2)

def g (x, y) :
    return x*x + 3*(y+1)**2 - 1

# Next their derivatives,
def dfdx (x, y) :
    return 1/2 * (-4*x + y) * f(x, y)

def dfdy (x, y) :
    return 1/2 * (x - 2*y) * f(x, y)

def dgdx (x, y) :
    return 2*x

def dgdy (x, y) :
    return 6*(y+1)
```

```python
from scipy import optimize

def DL (xyλ) :
    [x, y, λ] = xyλ
    return np.array([
            dfdx(x, y) - λ * dgdx(x, y),
            dfdy(x, y) - λ * dgdy(x, y),
            - g(x, y)
        ])

(x0, y0, λ0) = (-1, -1, 0)
# .x is an attribute of the object returned by the root-finding function
x, y, λ = optimize.root(DL, [x0, y0, λ0]).x
print("x = %g" % x)
print("y = %g" % y)
print("λ = %g" % λ)
print("f(x, y) = %g" % f(x, y))
```

```python
def find_all_stationary_points():
    # We'll store a list of dictionaries to keep guesses and results together
    detailed_results = []
    unique_roots = []
    
    x_range = np.linspace(-5, 5, 5) 
    y_range = np.linspace(-5, 5, 5)
    
    for x0 in x_range:
        for y0 in y_range:
            guess = [x0, y0, 0]
            res = optimize.root(DL, guess)
            
            if res.success:
                root = np.round(res.x, 6)
                
                # Check if this is a new unique root
                is_new = True
                for r in unique_roots:
                    if np.allclose(root, r, atol=1e-3):
                        is_new = False
                        break
                
                if is_new:
                    unique_roots.append(root)
                
                # Store the mapping of Guess -> Result
                detailed_results.append({
                    'guess': [x0, y0],
                    'root': root[:2], # Just x and y for the root
                    'f_val': f(root[0], root[1])
                })
    
    return detailed_results

# Execute
history = find_all_stationary_points()

print(f"{'Initial Guess (x, y)':<25} | {'Stationary Point (x, y)':<25} | {'f(x,y)':<10}")
print("-" * 75)
for item in history:
    g = item['guess']
    r = item['root']
    print(f"({g[0]:>5.1f}, {g[1]:>5.1f}){' ':10} | ({r[0]:>7.4f}, {r[1]:>7.4f}){' ':10} | {item['f_val']:>8.4f}")
```
